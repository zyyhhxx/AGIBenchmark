# 🧠 Cognitive Abilities Benchmark Suite
## Measuring What Matters for AGI: From Knowledge to Cognition

### Submission for the Kaggle "Measuring Progress Toward AGI" Hackathon

---

This suite measures **five core cognitive abilities** in language models, grounded in established cognitive science:

| # | Track | Benchmarks | Core Question |
|---|-------|------------|---------------|
| 1 | **Metacognition** | 9 | Does the model know what it knows? |
| 2 | **Learning** | 4 | Can it learn from examples and transfer? |
| 3 | **Attention** | 4 | Can it selectively process and sustain focus? |
| 4 | **Executive Functions** | 5 | Can it plan, adapt, and inhibit? |
| 5 | **Social Cognition** | 4 | Can it understand other minds? |

**29 benchmarks total** · All scores [0,1] · Contamination-resistant · Psychometrically validated

---

## Key Design Principles

1. **Cognitive science grounding** — Every benchmark maps to a named theoretical framework with human baselines
2. **Contamination resistance** — Procedurally generated stimuli, canary items, invented rule systems
3. **Shortcut resistance** — Two-phase protocols, adversarial items, multi-dimensional scoring
4. **Psychometric validation** — Cronbach's α ≥ 0.70, good discriminant validity (within-track r = 0.37 vs between-track r = 0.09)

## What We Measure That Nobody Else Does

- **Metacognitive monitoring** — calibration, FOK, JOL (not just accuracy)
- **Learning dynamics** — power-law curves, interference, transfer (not just few-shot)
- **Cognitive control** — set-shifting, planning, working memory updating
- **Genuine social understanding** — theory of mind, pragmatic inference
- **Self-monitoring under uncertainty** — canaries + epistemic revision
