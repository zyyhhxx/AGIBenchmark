## Cognitive Science Rationale

**Belief revision** is the ability to update knowledge when presented with contradictory evidence. Using the invented "Zorblatt Chemistry" rule system, this benchmark teaches 10 rules, then contradicts 3, measuring whether the model appropriately updates beliefs AND propagates changes to downstream inferences.


## Interpreting the Score

Score = accuracy on transfer questions after contradiction. 1.0 = perfect belief revision including downstream consequences.
